### Convertir la columna tipo str a num

```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"].str.strip(),
    errors="coerce" # "coerce" convierte los valores no convertibles en NaN
)
```

Imputar los valores NaN con 0 con la condicion de teneure en 0

```python
df.loc[
    (df["TotalCharges"].isna()) & (df["tenure"] == 0),
    "TotalCharges"
] = 0
```

## **Feature Engineering**

Feature 1: AvgMonthltyCharges

```python
# Feature 1: cantidad de servicios adicionales
service_cols = [
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

df["TotalServices"] = (
    df[service_cols]
    .eq("Yes")
    .sum(axis=1)
)
```

Feature 2: ChargeDifference

```python
# Feature 2: grupo de antigüedad
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=[
        "0-12 meses",
        "13-24 meses",
        "25-48 meses",
        "49-72 meses"
    ]
)
```
